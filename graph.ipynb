```python
#matplotlib line
import matplotlib.pyplot as plt
import numpy as np

#dashline
ypoint=np.array([3,8,1,10])
ypoint=np.array([3,8,1,10])

plt.plot(ypoint,linestyle='dashdot')
plt.show()

#line color
ypoint=np.array([3,8,1,10])

plt.plot(ypoint,color='y')
plt.show()

#multiple line
y1=np.array([3,8,1,10])
y2=np.array([6,2,7,11])

plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
#add grid line to a plot
import matplotlib.pyplot as plt
import numpy as np

ypoint=np.array([3,8,1,10])

plt.plot(ypoint)
plt.grid()
plt.show()

#display only x axis
ypoint=np.array([3,8,1,10])

plt.plot(ypoint)
plt.grid(axis='x')
plt.show()
```
