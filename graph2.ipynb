```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Example 1: Two Scatter Plots
# -----------------------------
x1 = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y1 = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

x2 = np.array([2,2,8,1,15,8,12,9,7,3,11,4,7,14,12])
y2 = np.array([100,105,84,105,90,99,90,95,94,100,79,112,91,80,85])

plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.scatter(x1, y1, color="blue")
plt.scatter(x2, y2, color="red")
plt.title("Two Scatter Plots")

# -----------------------------
# Example 2: Marker Size
# -----------------------------
sizes = np.array([20,50,100,200,500,1000,60,90,10,300,600,800,75])

plt.subplot(2,2,2)
plt.scatter(x1, y1, s=sizes, color="green")
plt.title("Marker Size")

# -----------------------------
# Example 3: Transparency
# -----------------------------
plt.subplot(2,2,3)
plt.scatter(x1, y1, s=sizes, alpha=0.5, color="orange")
plt.title("Alpha = 0.5")

# -----------------------------
# Example 4: Color Map
# -----------------------------
x = np.random.randint(100, size=100)
y = np.random.randint(100, size=100)
colors = np.random.randint(100, size=100)
sizes = 10 * np.random.randint(100, size=100)

plt.subplot(2,2,4)
scatter = plt.scatter(x, y, c=colors, s=sizes,
                      alpha=0.5, cmap="nipy_spectral")
plt.colorbar(scatter)
plt.title("Color Map")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

# Create a figure with 1 row and 4 columns
plt.figure(figsize=(16, 4))

# 1. Vertical Bar Chart
plt.subplot(1, 4, 1)
plt.bar(x, y)
plt.title("Vertical Bar")

# 2. Horizontal Bar Chart
plt.subplot(1, 4, 2)
plt.barh(x, y)
plt.title("Horizontal Bar")

# 3. Colored Bar Chart
plt.subplot(1, 4, 3)
plt.bar(x, y, color="green")
plt.title("Green Bar")

# 4. Narrow Width Bar Chart
plt.subplot(1, 4, 4)
plt.bar(x, y, width=0.1)
plt.title("Width = 0.1")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
y = np.array([35, 25, 25, 15])
labels = ["Apples", "Bananas", "Cherries", "Dates"]

# 1. Simple Pie Chart
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.pie(y)
plt.title("Simple Pie")

# 2. Pie Chart with Labels
plt.subplot(2, 3, 2)
plt.pie(y, labels=labels)
plt.title("With Labels")

# 3. Pie Chart with Start Angle
plt.subplot(2, 3, 3)
plt.pie(y, labels=labels, startangle=90)
plt.title("Start Angle = 90°")

# 4. Exploded Pie Chart
explode = [0, 0, 0.2, 0]
plt.subplot(2, 3, 4)
plt.pie(y, labels=labels, explode=explode)
plt.title("Exploded Slice")

# 5. Pie Chart with Percentage
plt.subplot(2, 3, 5)
plt.pie(y, labels=labels, autopct="%1.1f%%")
plt.title("With Percentage")

# 6. Pie Chart with Legend
plt.subplot(2, 3, 6)
plt.pie(y)
plt.legend(labels, title="Four Fruits")
plt.title("With Legend")

plt.tight_layout()
plt.show()
```
