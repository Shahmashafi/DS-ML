```python
import matplotlib.pyplot as plt
import numpy as np

#1
x=np.array([1,6])
y=np.array([2,5])
plt.subplot(2,3,1)
plt.plot(x,y,'o')
plt.title("Title 1")
plt.xlabel("x label",color="blue")
plt.ylabel("y label",color="blue")

#2
x1=np.array([1,2,3,4,5])
y1=np.array([1,4,2,3,4])
x2=np.array([1,2,3,4,5])
y2=np.array([2,2,3,4,5])
plt.subplot(2,3,2)
plt.plot(x1,y1,linestyle='dashed')
plt.plot(x2,y2,marker="*",linewidth=2.5)

#3
ypoint=np.array([3,8,1,10])
plt.subplot(2,3,3)
plt.plot(ypoint,marker='o',ms=10,mec='black',mfc='yellow')

#4
ypoint=np.array([0,1,2,3,4])
plt.subplot(2,3,4)
plt.title("Title 2")
plt.plot(ypoint)
plt.grid()

#5
x1=np.array([1,2,3,4,5,6,7,8,4,3])
y1=np.array([1,4,2,3,2,3,6,7,8,2])
colors=np.array(["red","pink","yellow","blue","green","magenta","lime","cyan","orange","teal"])
sizes=np.array([100,50,60,50,100,30,250,30,60,230])
plt.subplot(2,3,5)
plt.scatter(x1,y1,c=colors,s=sizes,alpha=0.5)

#6
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.subplot(2,3,6)
plt.bar(x,y,width=0.1,color="red")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([35,25,20,20])
exp=[0.2,0,0,0]
clr=["red","green","blue","yellow"]
lab=["A","B","c","D"]
plt.pie(x,startangle=90,explode=exp,shadow=True,colors=clr,labels=lab)
plt.legend()
plt.show()
```
