```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

# Sample dataset
texts = [
    "India won the cricket match",
    "The football team scored a goal",
    "The player hit a century",
    "The team won the championship",
    "The match was exciting",
    "The bowler took five wickets",
    "Python is used for machine learning",
    "Artificial intelligence is growing rapidly",
    "The computer uses a new processor",
    "Machine learning models need data",
    "Cloud computing provides online services",
    "The software application uses AI"
]

# Class labels
labels = [
    "Sports", "Sports", "Sports",
    "Sports", "Sports", "Sports",
    "Technology", "Technology", "Technology",
    "Technology", "Technology", "Technology"
]

# Convert text into numerical features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42
)

# Create SVM classifier
model = SVC(kernel="linear")

# Train the model
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')


print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Test with new text
new_text = ["The cricket player scored a century"]
new_text_vector = vectorizer.transform(new_text)

prediction = model.predict(new_text_vector)

print("New Text:", new_text[0])
print("Predicted Class:", prediction[0])
```

```python
# Import libraries
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# New dataset
reviews = [
    "The movie was amazing and entertaining",
    "I really enjoyed this wonderful movie",
    "The acting was excellent",
    "The story was boring and disappointing",
    "I hated this terrible movie",
    "The movie was very bad",
    "Amazing acting and great story",
    "The film was enjoyable and interesting",
    "Poor acting and boring story",
    "This was an excellent film",
    "The movie was disappointing",
    "I loved the performance"
]

# 1 = Positive, 0 = Negative
labels = [1, 1, 1, 0, 0, 0, 1, 1, 0, 1, 0, 1]

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    reviews,
    labels,
    test_size=0.25,
    random_state=42
)

# Convert text into TF-IDF features
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Create SVM classifier
svm_classifier = SVC(kernel='linear')

# Train the model
svm_classifier.fit(X_train_tfidf, y_train)

# Predict test data
y_pred = svm_classifier.predict(X_test_tfidf)

# Evaluate model
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Test a new review
new_review = ["The movie was fantastic and enjoy"]

new_review_tfidf = vectorizer.transform(new_review)
print("New Text :",new_review[0])

prediction = svm_classifier.predict(new_review_tfidf)
print("Predicted Class :",prediction[0])
if prediction[0] == 1:
    print("Prediction: Positive Review")
else:
    print("Prediction: Negative Review")
```
